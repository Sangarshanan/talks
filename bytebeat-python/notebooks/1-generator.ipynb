# 🎵 Bytebeat Music Generator

Generate music from simple mathematical expressions! This notebook has the generator and examples.

Bytebeat works by applying bitwise operations to a time counter `t` that increments with each audio sample.

```python
import struct
import io
import wave
from IPython.display import Audio

class BytebeatPlayer:
    """Bytebeat generator optimized for iPython/Jupyter playback"""
    
    def __init__(self, sample_rate=8000):
        self.sample_rate = sample_rate
    
    def generate(self, formula, duration=10):
        """Generate bytebeat samples from a formula"""
        num_samples = int(duration * self.sample_rate)
        samples = []
        
        for t in range(num_samples):
            try:
                value = int(formula(t)) & 0xFF  # Keep in 0-255 range
                samples.append(value)
            except:
                samples.append(128)  # Silence on error
        
        return samples
    
    def play(self, formula, duration=10):
        """Generate and play audio directly in iPython"""
        samples = self.generate(formula, duration)
        
        # Convert 8-bit samples (0-255) to 16-bit signed (-32768 to 32767)
        samples_16bit = [(s - 128) * 256 for s in samples]
        
        # Create WAV data in memory
        byte_io = io.BytesIO()
        
        with wave.open(byte_io, 'wb') as wav:
            wav.setnchannels(1)  # Mono
            wav.setsampwidth(2)  # 16-bit
            wav.setframerate(self.sample_rate)
            wav.writeframes(struct.pack(f'{len(samples_16bit)}h', *samples_16bit))
        
        byte_io.seek(0)
        return Audio(data=byte_io.read(), rate=self.sample_rate, autoplay=False)

# Initialize player
bb = BytebeatPlayer(sample_rate=8000)
```

## 🎼 Some Bytebeat Formulas

Run each cell to hear them!

```python
# Classic 1 - The original bytebeat formula by Viznut (2011)
bb.play(
    lambda t: t * ((t >> 12 | t >> 8) & 63 & t >> 4), duration=15
)
```

```python
# Classic 2 - Another rhythmic pattern
bb.play(
    lambda t: t * (t >> 8 | t >> 9 | t >> 13 | t >> 5), duration=15
)
```

```python
# Sierpinski Triangle
bb.play(lambda t: t & (t >> 8), duration=15)
```

```python
# lot of bit manipulation
bb.play(lambda t: (t >> 6 | t << 1) + (t >> 5 | t << 3 | t >> 3) | t >> 2 | t, duration=20)
```

```python
# Arpeggio?
bb.play(lambda t: t * ((t >> 11 & t >> 8) & 123 & (t >> 3)), duration=15)
```

```python
# Chaos
bb.play(lambda t: (t | t >> 9 | t >> 7) * t & (t >> 11 | t >> 9), duration=15)
```

## 💡 Make your own?

1. **Start simple**: Begin with `t >> 6` and build up
2. **Add rhythm**: Use `&` to create rhythmic patterns
3. **Layer sounds**: Use `|` to combine different patterns
4. **Multiply for pitch**: Try `t * 2` or `t * 3`
5. **Combine operations**: Mix everything together!
