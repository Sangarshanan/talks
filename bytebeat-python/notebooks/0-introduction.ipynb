# Time (t)

How is audio represented on computers?

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio

sample_rate = 8000
duration = 1
    
num_samples = int(duration * sample_rate) # total number of samples
t = np.arange(num_samples) # array of sample indices
samples = (t % 256) # cycle length of 256 samples

plt.plot(samples)
Audio(data=samples, rate=sample_rate, autoplay=False)
```

# Plug in a Formula

Modify our array with a function

```python
sample_rate = 8000
duration = 0.5

num_samples = int(duration * sample_rate)
t = np.arange(num_samples)

formula = lambda t: (t * 3)

samples = formula(t % 256).astype(np.uint8)

plt.plot(t[:10000], samples[:10000])
Audio(data=samples, rate=sample_rate, autoplay=False)
```

# More Operations

```python
sample_rate = 8000
duration = 15

num_samples = int(duration * sample_rate)
t = np.arange(num_samples)

formula = lambda t: (t&t<<6&t>>7|-t>>4|t&t>>10)

samples = formula(t).astype(np.uint8)

plt.plot(t[:40000], samples[:40000])
Audio(data=samples, rate=sample_rate, autoplay=False)
```
